## 1. Override inherited behavior


```python
class Person:
    def introduce(self):
        print("I am a person")

class Doctor(Person):
    def introduce(self):
        print("I am a doctor")

person = Person()
doctor = Doctor()
person.introduce()
doctor.introduce()
```

## 2. Extend superclass behavior with `super()`


```python
class Person:
    def introduce(self):
        print("I am a person")

class Doctor(Person):
    def introduce(self):
        super().introduce()
        print("My role is doctor")

Doctor().introduce()
```

## 3. Polymorphism

One operation name, different implementations.

```python
class Notification:
    def send(self):
        print("Sending generic notification")

class EmailNotification(Notification):
    def send(self):
        print("Sending email")

class SMSNotification(Notification):
    def send(self):
        print("Sending SMS")

class PushNotification(Notification):
    def send(self):
        print("Sending push notification")

notifications = [EmailNotification(), SMSNotification(), PushNotification()]

for notification in notifications:
    notification.send()
```

### Think

The loop does not check the concrete type. What determines which `send()` implementation runs?

## 4. Dynamic binding

Predict the output before running the next cell.

```python
class Person:
    def role(self):
        print("Person")

class Doctor(Person):
    def role(self):
        print("Doctor")

class Patient(Person):
    def role(self):
        print("Patient")

people = [Person(), Doctor(), Patient()]
for person in people:
    person.role()
```
