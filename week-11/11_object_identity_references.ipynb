## 1. Two names, one object


```python
class Student:
    def __init__(self, name):
        self.name = name

student_a = Student("Alya")
student_b = student_a

print(student_a is student_b)
print(id(student_a))
print(id(student_b))
```

### Think

How many Student objects exist? How many variable names refer to that object?

## 2. Identity vs equality


```python
a = [1, 2, 3]
b = [1, 2, 3]

print("a == b:", a == b)
print("a is b:", a is b)
```

`==` asks about value equality. `is` asks whether both references point to the same object.

## 3. Aliasing and mutation


```python
x = [10, 20]
y = x

y.append(30)

print("x:", x)
print("y:", y)
print("same object:", x is y)
```

## 4. Reassignment is different from mutation


```python
x = [1, 2]
y = x

y.append(3)
print("after mutation")
print("x:", x)
print("y:", y)
print("x is y:", x is y)

y = [100, 200]
print("after reassignment")
print("x:", x)
print("y:", y)
print("x is y:", x is y)
```

## 5. Aliasing with custom objects


```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

account_a = BankAccount(100_000)
account_b = account_a

account_b.deposit(50_000)

print(account_a.balance)
print(account_b.balance)
print(account_a is account_b)
```

## 6. Reference lifetime

The object can remain reachable through another reference even after one name is rebound.

```python
student_a = Student("Alya")
student_b = student_a

student_a = None

print(student_a)
print(student_b.name)
```
