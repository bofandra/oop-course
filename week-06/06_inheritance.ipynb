## 1. Start with the is-a test

Before using inheritance, ask whether the specialized concept really **is a** kind of the general concept.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print(self.name)

class Doctor(Person):
    pass

doctor = Doctor("Dr. Andi")
doctor.display_name()
```

### Think

Why does `Doctor(Person)` make conceptual sense, while `Car(Engine)` usually does not?

## 2. Add subclass-specific state and behavior


```python
class Person:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print(self.name)

class Doctor(Person):
    def __init__(self, name, specialty):
        super().__init__(name)
        self.specialty = specialty

    def diagnose(self):
        print(self.name, "is diagnosing a patient")

doctor = Doctor("Dr. Andi", "Cardiology")
doctor.display_name()
doctor.diagnose()
print(doctor.specialty)
```

## 3. Multiple subclasses of one superclass


```python
class Patient(Person):
    def __init__(self, name, patient_id):
        super().__init__(name)
        self.patient_id = patient_id

    def show_patient_id(self):
        print(self.patient_id)

patient = Patient("Budi", "P001")
patient.display_name()
patient.show_patient_id()
```

## 4. Check inherited access

A subclass instance can use inherited superclass behavior plus subclass-specific behavior.

```python
print(isinstance(doctor, Doctor))
print(isinstance(doctor, Person))
print(isinstance(patient, Person))
```
