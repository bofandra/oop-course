## 1. The problem with uncontrolled state

Run the example and observe that outside code can assign any value.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

account = BankAccount("Alya", 100_000)
account.balance = -1_000_000
print(account.balance)
````

### Think

- Is this state valid for our simple domain model?
- Which object should control changes to balance?

## 2. Internal-use state and a public interface


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount

account = BankAccount("Alya", 100_000)
account.deposit(50_000)
account.withdraw(25_000)
print(account._balance)
````

A leading underscore is a **Python convention**, not strict privacy. The normal interface is `deposit()` / `withdraw()` rather than direct mutation of `_balance`.

## 3. Read-only access with `@property`


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self._balance:
            self._balance -= amount

account = BankAccount("Alya", 100_000)
print(account.balance)
````
