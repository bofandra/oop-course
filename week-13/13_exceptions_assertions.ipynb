## 1. A failure without protection


````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        self.balance -= amount

account = BankAccount(100_000)
account.withdraw(500_000)
print(account.balance)
````

### Think

If the domain invariant is `balance >= 0`, what went wrong?

## 2. Raise a Python exception


````python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative")
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient balance")
        self._balance -= amount
        assert self._balance >= 0
````

## 3. Handle an expected exception


````python
account = BankAccount(100_000)

try:
    account.withdraw(500_000)
except ValueError as error:
    print("Withdrawal failed:", error)

print("Balance:", account.balance)
````

## 4. Exception propagation


````python
def pay_bill(account, amount):
    account.withdraw(amount)

try:
    pay_bill(account, 500_000)
except ValueError as error:
    print("Handled by caller:", error)
````

## 5. Preconditions, postconditions, invariant

For `withdraw(amount)` write the contract before coding:

```text
Preconditions:
- amount > 0
- amount <= balance

Postcondition:
- new balance = old balance - amount

Invariant:
- balance >= 0
```

## 6. Transition rule as a precondition

A workflow rule can depend on the object's current state. For a Rental lifecycle `created → active → completed`, `complete()` has a state-dependent precondition: `status == "active"`.

The invariant is different: status must always remain one of the valid status values.

````python
class Rental:
    VALID_STATUSES = {"created", "active", "completed"}

    def __init__(self):
        self.status = "created"

    def start(self):
        if self.status != "created":
            raise ValueError("Only a created rental can be started")
        self.status = "active"
        assert self.status in self.VALID_STATUSES

    def complete(self):
        if self.status != "active":
            raise ValueError("Only an active rental can be completed")
        self.status = "completed"
        assert self.status in self.VALID_STATUSES

rental = Rental()

try:
    rental.complete()
except ValueError as error:
    print("Completion rejected:", error)

rental.start()
rental.complete()
print(rental.status)
````
