## 1. `__str__()` — readable object representation


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

p = Point(2, 3)
print(p)
```

## 2. `__eq__()` — value equality vs identity


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

p1 = Point(1, 2)
p2 = Point(1, 2)

print("same identity:", p1 is p2)
print("equal value:", p1 == p2)
```

For this introductory `__eq__()` example, assume the other operand is also a `Point`. Handling comparisons with unrelated types is not required in this module.

## 3. `__add__()` — operator overloading


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)
```

For this introductory `__add__()` example, assume both operands are `Point` objects. Focus on giving `+` a meaningful domain interpretation.

## 6. Module organization

This repository also contains `models.py` and `main.py`. In Colab, you can recreate the same idea by writing a module file.

```python
module_code = '''
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
'''

with open('models.py', 'w') as file:
    file.write(module_code)
```
