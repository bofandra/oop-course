## 2. Initializing instance attributes

`__init__()` initializes the state of each newly created instance.

```python
class Patient:
    def __init__(self, name, age):
        self.name = name
        self.age = age

patient_1 = Patient("Budi", 30)
patient_2 = Patient("Siti", 25)

print(patient_1.name, patient_1.age)
print(patient_2.name, patient_2.age)
```

### Observe `self`

The same class definition works with different instances. Inside an instance method, `self` refers to the current instance.

```python
class Patient:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(self.name, self.age)

patient_1 = Patient("Budi", 30)
patient_2 = Patient("Siti", 25)

patient_1.display_info()
patient_2.display_info()
```

## 3. Instance attributes vs class attributes


```python
class Student:
    university = "Example University"

    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name

student_1 = Student("S001", "Alya")
student_2 = Student("S002", "Bima")

print(Student.university)
print(student_1.student_id, student_1.name, student_1.university)
print(student_2.student_id, student_2.name, student_2.university)
```

### Classify

Which attributes belong to each individual student, and which information is shared at class level?
