## 1. Variables and expressions


```python
name = "Alya"
age = 19
credits = 4

print(name)
print(age + 1)
print(credits * 2)
```

## 2. Conditionals


```python
score = 78

if score >= 85:
    print("A")
elif score >= 70:
    print("B")
else:
    print("Review needed")
```

## 3. Loops


```python
students = ["Alya", "Budi", "Siti"]

for student in students:
    print(student)
```

## 4. Functions


```python
def greet(name):
    return f"Hello, {name}"

for student in students:
    print(greet(student))
```
