## 1. Appointment refers to Patient and Doctor


```python
class Patient:
    def __init__(self, patient_id, name):
        self.patient_id = patient_id
        self.name = name

class Doctor:
    def __init__(self, doctor_id, name):
        self.doctor_id = doctor_id
        self.name = name

class Appointment:
    def __init__(self, patient, doctor, time):
        self.patient = patient
        self.doctor = doctor
        self.time = time
        self.status = "waiting"

patient = Patient("P001", "Budi")
doctor = Doctor("D001", "Dr. Andi")
appointment = Appointment(patient, doctor, "10:00")

print(appointment.patient.name)
print(appointment.doctor.name)
```

### Think

Why is storing the Patient object different from storing only `patient_name`?

### A reference is not automatically composition

`Appointment` referring to `Patient` is an association/reference relationship. The Patient has an identity and lifecycle outside the Appointment.

Reserve whole–part/composition-style language for relationships where that meaning is actually useful, such as `Order` containing `OrderItem`.

## 2. Client–Supplier

One object can use a service provided by another object.

```python
class NotificationService:
    def send(self, message):
        print("Notification:", message)

class SimpleAppointment:
    def __init__(self):
        self.status = "waiting"

    def confirm(self, notification_service):
        self.status = "confirmed"
        notification_service.send("Appointment confirmed")

notification_service = NotificationService()
appointment = SimpleAppointment()
appointment.confirm(notification_service)
```

## 3. Has-a relationship


```python
class Engine:
    def __init__(self, model):
        self.model = model

class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine

engine = Engine("E-100")
car = Car("Example", engine)
print(car.brand, car.engine.model)
```

## 4. Composition — Order, OrderItem, Product


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class OrderItem:
    def __init__(self, product, quantity):
        self.product = product
        self.quantity = quantity

    def subtotal(self):
        return self.product.price * self.quantity

class Order:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def total(self):
        return sum(item.subtotal() for item in self.items)

keyboard = Product("Keyboard", 500_000)
mouse = Product("Mouse", 200_000)
order = Order()
order.add_item(OrderItem(keyboard, 1))
order.add_item(OrderItem(mouse, 2))
print(order.total())
```
