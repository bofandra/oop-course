## 1. Why a normal superclass may be too weak

A generic Shape returning `0` for area technically runs, but it does not represent a meaningful concrete shape.

```python
class Shape:
    def area(self):
        return 0

shape = Shape()
print(shape.area())
```

## 2. Python implementation bridge: abstract base class


```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

Try to instantiate `Shape()` and observe what Python does.

```python
# Uncomment to experiment:
# shape = Shape()
```

## 3. Concrete subclass implements required behavior


```python
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rectangle = Rectangle(4, 5)
print(rectangle.area())
```

## 4. Abstract class can also contain concrete/shared behavior


```python
class Employee(ABC):
    def __init__(self, employee_id, name):
        self.employee_id = employee_id
        self.name = name

    def display_name(self):
        return self.name

    @abstractmethod
    def calculate_pay(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, employee_id, name, salary):
        super().__init__(employee_id, name)
        self.salary = salary

    def calculate_pay(self):
        return self.salary

employee = FullTimeEmployee("E001", "Alya", 8_000_000)
print(employee.display_name())
print(employee.calculate_pay())
```
